```python
import pandas as pd
import numpy as np

# -------------------------
# Cargar dataset
# -------------------------
df = pd.read_csv("data/raw/dataset3_red_vial.csv", encoding="latin-1")

print("ANTES DE LIMPIEZA")
print("Filas:", len(df))
print("Nulos:\n", df.isnull().sum())
print("Duplicados:", df.duplicated().sum())

# -------------------------
# 1. Eliminar duplicados
# -------------------------
df = df.drop_duplicates()

# -------------------------
# 2. Tratamiento de nulos
# -------------------------
# Ejemplo:
# - columnas numéricas → imputar con media
# - columnas categóricas → imputar con "Desconocido"

for col in df.select_dtypes(include=np.number):
 df[col] = df[col].fillna(df[col].mean())

for col in df.select_dtypes(include='object'):
 df[col] = df[col].fillna("Desconocido")

# -------------------------
# 3. Conversión de tipos
# -------------------------
# (ejemplo: convertir a numérico si hay errores)

for col in df.columns:
    try:
        df[col] = pd.to_numeric(df[col])
    except:
        pass

# -------------------------
# 4. Normalización de texto
# -------------------------
for col in df.select_dtypes(include='object'):
 df[col] = df[col].str.strip().str.lower()

# -------------------------
# 5. Tratamiento de outliers (IQR)
# -------------------------

numeric_cols = df.select_dtypes(include=np.number).columns

Q1 = df[numeric_cols].quantile(0.25)
Q3 = df[numeric_cols].quantile(0.75)
IQR = Q3 - Q1

# Limitar valores extremos solo en columnas numéricas
mask = ~((df[numeric_cols] < (Q1 - 1.5 * IQR)) | (df[numeric_cols] > (Q3 + 1.5 * IQR))).any(axis=1)
df = df[mask]
# -------------------------
# 6. Guardar dataset limpio
# -------------------------
df.to_csv("data/raw/dataset3_limpio.csv", index=False)

# -------------------------
# DESPUÉS DE LIMPIEZA
# -------------------------
print("\nDESPUÉS DE LIMPIEZA")
print("Filas:", len(df))
print("Nulos:\n", df.isnull().sum())
print("Duplicados:", df.duplicated().sum())
```
