```python
import pandas as pd
from scipy.stats import ttest_ind
df = pd.read_csv("data/raw/dataset3_limpio.csv")
lima = df[df['DEPARTAMENTO'] == 'lima']['VEH_TOTAL']
otras = df[df['DEPARTAMENTO'] != 'lima']['VEH_TOTAL']
t_stat, p_value = ttest_ind(lima, otras, equal_var=False)
print("Estadístico t:", t_stat)
print("p-value:", p_value)
alpha = 0.05
if p_value < alpha:
    print("Se rechaza H0")
else:
    print("No se rechaza H0")
```
